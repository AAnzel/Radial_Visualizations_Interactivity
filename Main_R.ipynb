```r
library(ggplot2)

df <- data.frame(
  Feature=c("sepal length","sepal width","petal length","petal width"),
  Value=c(0.222, 1, 0.0667, 0.0417)
)
df$ymax <- cumsum(df$Value)
df$ymin <- c(0, head(df$ymax, n=-1))
pdf("Figures/ggplot2_pie.pdf", width=6, height=6)
ggplot(df, aes(ymax=ymax, ymin=ymin, xmax=4, xmin=3, fill=Feature)) +
  geom_rect() +
  coord_polar(theta="y") +
  theme_void(base_size=15) +
  ggtitle("Iris Setosa Feature Distribution")
dev.off()
```

```r
library(ggplot2)
library(dplyr)

# Data
df <- data.frame(
  Feature = c("sepal length", "sepal width", "petal length", "petal width"),
  Value = c(0.222, 1.0, 0.0667, 0.0417)
)

# Assign angle (evenly spaced)
df <- df %>%
  mutate(
    angle = seq(0, 2 * pi, length.out = n() + 1)[- (n() + 1)],  # skip last repeated point
    x = Value * sin(angle),
    y = Value * cos(angle)
  )

# Close the polygon by repeating the first row
df_closed <- bind_rows(df, df[1, ])

# Create radial axes and labels
axes <- df %>%
  mutate(x_end = sin(angle), y_end = cos(angle))

# Output directory
if (!dir.exists("Figures")) dir.create("Figures")
pdf("Figures/ggplot2_radar.pdf", width = 6, height = 6)

# Plot
ggplot() +
  # Radar polygon
  geom_polygon(data = df_closed, aes(x = x, y = y), fill = "firebrick", alpha = 0.3, color = "firebrick") +
  geom_point(data = df, aes(x = x, y = y), color = "firebrick", size = 3) +
  # Axes
  geom_segment(data = axes, aes(x = 0, y = 0, xend = x_end, yend = y_end), color = "gray80") +
  geom_text(data = axes, aes(x = 1.15 * sin(angle), y = 1.15 * cos(angle), label = Feature), size = 4) +
  # Circle guides
  annotate("path", x = cos(seq(0, 2 * pi, length.out = 100)) * 0.5, y = sin(seq(0, 2 * pi, length.out = 100)) * 0.5, color = "grey90") +
  annotate("path", x = cos(seq(0, 2 * pi, length.out = 100)) * 1.0, y = sin(seq(0, 2 * pi, length.out = 100)) * 1.0, color = "grey90") +
  coord_equal() +
  labs(title = "Iris Setosa Feature Profile") +
  theme_minimal(base_size = 15) +
  theme(axis.text = element_blank(), axis.title = element_blank(), panel.grid = element_blank(), axis.ticks = element_blank())

dev.off()
```

```r
df <- data.frame(
  Feature=factor(c("sepal length","sepal width","petal length","petal width"), 
                 levels=c("sepal length","sepal width","petal length","petal width")),
  Value=c(0.222, 1, 0.0667, 0.0417)
)
pdf("Figures/ggplot2_polarbar.pdf", width=6, height=6)
ggplot(df, aes(x=Feature, y=Value, fill=Feature)) +
  geom_bar(stat="identity", width=1) +
  ylim(0,1) +
  coord_polar() +
  labs(title="Iris Setosa Feature: Polar Bar Chart", y=NULL, x=NULL) +
  theme_minimal(base_size=15)
dev.off()
```
