```python
import os
import numpy as np
import pandas as pd
from sklearn.metrics import cohen_kappa_score

_PATH_CODING_DIR = os.path.join('..', 'Data', 'Coding_Results')
_PATH_CODING_TABLE_1 = os.path.join(_PATH_CODING_DIR, 'table1_coding.csv')
_PATH_CODING_TABLE_2 = os.path.join(_PATH_CODING_DIR, 'table2_coding.csv')
```

# Table 1 --- Agreement Analysis

```python
df_table_1 = pd.read_csv(_PATH_CODING_TABLE_1)
df_table_1
```

```python
dict_label_mapping = {"Full": 2, "Partial": 1, np.nan: 0}

df_table_1["coder1_int"] = df_table_1["coder1_support"].map(dict_label_mapping)
df_table_1["coder2_int"] = df_table_1["coder2_support"].map(dict_label_mapping)

if df_table_1["coder1_int"].isnull().any() or df_table_1["coder2_int"].isnull().any():
    missing_rows = df_table_1[df_table_1["coder1_int"].isnull() | df_table_1["coder2_int"].isnull()]
    raise ValueError(
        "Found unsupported labels in coder1_support/coder2_support.\n"
        f"Problematic rows:\n{missing_rows}"
    )

df_table_1
```

```python
kappa_overall = cohen_kappa_score(df_table_1["coder1_int"], df_table_1["coder2_int"])
print(f"Overall Cohen's kappa across all triples: {kappa_overall:.3f}")
```

```python
# Compute kappa per interaction category
for cat in df_table_1["interaction_category"].unique():
    subset = df_table_1[df_table_1["interaction_category"] == cat]
    if len(subset) < 2:
        continue
    kappa_cat = cohen_kappa_score(subset["coder1_int"], subset["coder2_int"])
    print(
        f"Cohen's kappa for interaction_category='{cat}' "
        f"({len(subset)} triples): {kappa_cat:.3f}"
    )
```
